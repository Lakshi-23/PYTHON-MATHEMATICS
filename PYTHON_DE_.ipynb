```python
#  Write a Python program to visualize the direction field for dy/dx = x - y.
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=X-Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD: Dy/Dx=x-y")
plt.show()
```

```python
#Develop a Python program to plot the slope field for  dy/dx = xy.
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=X*Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD; dy/dx = xy")
plt.show()
```

```python
#Write a Python code to plot the slope field for  dy/dx = sin(x) - y.
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=np.sin(X)-Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD; dy/dx=sinx -y ")
plt.show()
```

```python
#Solve the differential equation using Python and plot the solution  dy/dx + y = e^x using Python.
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=np.exp(X)-Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD:dy/dx = e^x - y ")
plt.show()
```

```python
#Solve the following differential equation using Python  dy/dx - 2y = 6.
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=6+2*Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD: dy/dx = 6 +2y")
plt.show()
```

```python
#Solve and plot the solution of the differential equation using Python  dy/dx + 2y = cos(x).
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-5,5,20)
y=np.linspace(-5,5,20)
X,Y=np.meshgrid(x,y)
M=np.cos(X)-2*Y
U=1/np.sqrt(1+M**2)
V=M/np.sqrt(1+M**2)
plt.quiver(X,Y,U,V)
plt.title("SLOPE FIELD: dy/dx = cos(x)-2y")
plt.show()
```

```python
#Convert y'' + y = 0 into a system of first order equations and plot the solution using python.
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
def system(t,z):
    y,v=z
    return [v,-y]
sol=solve_ivp(system,[0,10],[1,0],t_eval=[i/10 for i in range(101)])
plt.plot(sol.t,sol.y[0])
plt.title("y'' + y = 0 ")
plt.show()

```

```python
#Using Python, solve and visualize the damped oscillator equation y'' + 5y' + 4y = 0.
from scipy.integrate import solve_ivp
import matplotlib.pyplot as plt
def system(t,z):
    y,v=z
    return [v,-5*v-4*y]
sol=solve_ivp(system,[0,10],[1,0],t_eval=[i/10 for i in range(101)])
plt.plot(sol.t,sol.y[0])
plt.title("y'' + 5y' + 4y = 0 ")
plt.show()
```
